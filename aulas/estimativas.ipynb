```python
import math
from scipy.stats import norm, t


def intervalo_confianca_media(
    media_amostral,
    desvio_amostral,
    n,
    confianca=0.95,
    populacao_finita=False,
    N=None
):

    if n <= 1:
        raise ValueError('n deve ser maior que 1')

    if desvio_amostral < 0:
        raise ValueError('O desvio-padrão amostral não pode ser negativo')

    if not (0 < confianca < 1):
        raise ValueError('A confiança deve estar entre 0 e 1. Ex: 0.95')

    alpha = 1 - confianca
    gl = n - 1

    # Escolha do método
    if n < 30:
        metodo = 't de Student'
        valor_critico = t.ppf(1 - alpha / 2, df=gl)
    else:
        metodo = 'normal z'
        valor_critico = norm.ppf(1 - alpha / 2)

    # Erro padrão
    erro_padrao = desvio_amostral / math.sqrt(n)

    # Correção para população finita
    fpc = 1.0

    if populacao_finita:

        if N is None:
            raise ValueError('Informe N quando a população for finita')

        if N <= 1:
            raise ValueError('N deve ser maior que 1')

        if n > N:
            raise ValueError('n não pode ser maior que N')

        fpc = math.sqrt((N - n) / (N - 1))
        erro_padrao *= fpc

    # Intervalo
    margem_erro = valor_critico * erro_padrao

    li = media_amostral - margem_erro
    ls = media_amostral + margem_erro

    return {
        'metodo': metodo,
        'graus_liberdade': gl,
        'valor_critico': valor_critico,
        'erro_padrao': erro_padrao,
        'fator_correcao_pop_finita': fpc,
        'margem_erro': margem_erro,
        'limite_inferior': li,
        'limite_superior': ls,
    }
```

Calculo com poucas amostras

```python
r1 = intervalo_confianca_media(
    media_amostral= 0.4,
    desvio_amostral= 2.0,
    n = 25,
    confianca=0.99,
    populacao_finita=False
)

for k, v in r1.items():
    print(f'{k.title()}: {v}')
```

```python
import pandas as pd
import numpy as np

cam = r'C:\projetos_python\estatistica_v2\base_analitica_final.csv'
df_proj = pd.read_csv(cam)
df_proj['valorPagoEmpenho'] = df_proj['valorPagoEmpenho'].fillna(0)

serie = df_proj['valorPagoEmpenho']

N_pop   = len(serie)           # tamanho total da base (população finita)
media   = serie.mean()
desvio  = serie.std(ddof=1)

print(f'População total (N): {N_pop}')
print(f'Média real da base:  R$ {media:,.2f}')
print(f'Desvio-padrão:       R$ {desvio:,.2f}')
print()

# ── Cenário 1: dataset inteiro como amostra (Normal Z, n >> 30) ──────────────
r_total = intervalo_confianca_media(
    media_amostral=media,
    desvio_amostral=desvio,
    n=N_pop,
    confianca=0.95
)
print('--- Dataset completo como amostra (95%) ---')
for k, v in r_total.items():
    print(f'  {k.title()}: {v}')

print()

# ── Cenário 2: amostra aleatória de 200 linhas com correção pop. finita ──────
np.random.seed(42)
amostra = serie.sample(n=200)

r_amostra = intervalo_confianca_media(
    media_amostral=amostra.mean(),
    desvio_amostral=amostra.std(ddof=1),
    n=200,
    confianca=0.95,
    populacao_finita=True,
    N=N_pop
)
print('--- Amostra aleatória n=200, com correção pop. finita (95%) ---')
for k, v in r_amostra.items():
    print(f'  {k.title()}: {v}')

print()
print(f'  Média real da base está no intervalo? '
      f'{r_amostra["limite_inferior"] <= media <= r_amostra["limite_superior"]}')
```
